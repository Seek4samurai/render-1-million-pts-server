```python
# Size calculations
import os
import gc
print(os.path.getsize("../dataset/xl/xl.csv") / 1e6, "MB")
print(os.path.getsize("../dataset/xl/xl.parquet") / 1e6, "MB")
print(os.path.getsize("../dataset/xl/downcasted_xl.parquet") / 1e6, "MB")
gc.collect()
```

```python
# CSV loading metrics
import time
import pandas as pd
import gc

for i in range(5):
    start = time.time()
    df = pd.read_csv("../dataset/xl/xl.csv")
    load_time = time.time() - start
    
    print(f"Run {i+1} - Loading CSV:", load_time)
    print("DF Memory:", df.memory_usage(deep=True).sum() / 1024**2, "MB")
    
    del df
    gc.collect()
```

```python
# Parquet loading metrics
import time
import pandas as pd
import gc

for i in range(5):
    start = time.time()
    df = pd.read_parquet("../dataset/xl/xl.parquet")
    load_time = time.time() - start
    
    print(f"Run {i+1} - Loading Parquet:", load_time)
    print("DF Memory:", df.memory_usage(deep=True).sum() / 1024**2, "MB")
    
    del df
    gc.collect()
```

```python
# Parquet loading metrics with Arrow backend
import time
import pandas as pd
import gc
import pyarrow.parquet as pq

for i in range(5):
    start = time.time()
    df = pd.read_parquet("../dataset/xl/downcasted_xl.parquet", dtype_backend="pyarrow")
    load_time = time.time() - start
    
    print(f"Run {i+1} - Loading Parquet:", load_time)
    print("DF Memory:", df.memory_usage(deep=True).sum() / 1024**2, "MB")
    
    del df
    gc.collect()
```
